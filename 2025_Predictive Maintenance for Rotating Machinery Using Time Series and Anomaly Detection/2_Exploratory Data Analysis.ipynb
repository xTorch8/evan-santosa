# Exploratory Data Analysis

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import os
```

```python
DATASET_PATH = "../data/1st_test/test_1_clean.csv"
```

```python
df = pd.read_csv(DATASET_PATH)
df.head()
```

```python
df.info()
```

```python
df.describe().T
```

```python
fig, axes = plt.subplots(6, 1, figsize = (15, 30))

axes[0].plot(df["bearing_1_(x)_mean_absolute"])
axes[0].plot(df["bearing_2_(x)_mean_absolute"])
axes[0].plot(df["bearing_3_(x)_mean_absolute"])
axes[0].plot(df["bearing_4_(x)_mean_absolute"])
axes[0].legend(["Bearing 1", "Bearing 2", "Bearing 3", "Bearing 4"])
axes[0].set_title("Mean Absolute")

axes[1].plot(df["bearing_1_(x)_std"])
axes[1].plot(df["bearing_2_(x)_std"])
axes[1].plot(df["bearing_3_(x)_std"])
axes[1].plot(df["bearing_4_(x)_std"])
axes[1].legend(["Bearing 1", "Bearing 2", "Bearing 3", "Bearing 4"])
axes[1].set_title("STD")

axes[2].plot(df["bearing_1_(x)_skewness"])
axes[2].plot(df["bearing_2_(x)_skewness"])
axes[2].plot(df["bearing_3_(x)_skewness"])
axes[2].plot(df["bearing_4_(x)_skewness"])
axes[2].legend(["Bearing 1", "Bearing 2", "Bearing 3", "Bearing 4"])
axes[2].set_title("Skewness")

axes[3].plot(df["bearing_1_(x)_kurtosis"])
axes[3].plot(df["bearing_2_(x)_kurtosis"])
axes[3].plot(df["bearing_3_(x)_kurtosis"])
axes[3].plot(df["bearing_4_(x)_kurtosis"])
axes[3].legend(["Bearing 1", "Bearing 2", "Bearing 3", "Bearing 4"])
axes[3].set_title("Kurtosis")

axes[4].plot(df["bearing_1_(x)_rms"])
axes[4].plot(df["bearing_2_(x)_rms"])
axes[4].plot(df["bearing_3_(x)_rms"])
axes[4].plot(df["bearing_4_(x)_rms"])
axes[4].legend(["Bearing 1", "Bearing 2", "Bearing 3", "Bearing 4"])
axes[4].set_title("RMS")

axes[5].plot(df["bearing_1_(x)_crest_factor"])
axes[5].plot(df["bearing_2_(x)_crest_factor"])
axes[5].plot(df["bearing_3_(x)_crest_factor"])
axes[5].plot(df["bearing_4_(x)_crest_factor"])
axes[5].legend(["Bearing 1", "Bearing 2", "Bearing 3", "Bearing 4"])
axes[5].set_title("Crest Factor")

fig.show()
```

```python
correlation_pairs = df.drop(columns = ["timestamp"]).corr().abs().unstack().sort_values(ascending = False)
correlation_pairs = correlation_pairs[correlation_pairs < 1]
correlation_pairs.head(10)
```

```python
df["bearing_3_(x)_kurtosis_ma"] = df["bearing_3_(x)_kurtosis"].rolling(window = 50).mean()
df["bearing_4_(x)_kurtosis_ma"] = df["bearing_4_(x)_kurtosis"].rolling(window = 50).mean()
```

```python
fig, axes = plt.subplots(2, 1, figsize = (15, 10))

axes[0].plot(df["bearing_3_(x)_kurtosis"])
axes[0].plot(df["bearing_3_(x)_kurtosis_ma"])
axes[0].legend(["Kurtosis", "Moving Average (50)"])
axes[0].set_title("Kurtosis Bearing 3")

axes[1].plot(df["bearing_4_(x)_kurtosis"])
axes[1].plot(df["bearing_4_(x)_kurtosis_ma"])
axes[1].legend(["Kurtosis", "Moving Average (50)"])
axes[1].set_title("Kurtosis Bearing 4")

fig.show()
```
